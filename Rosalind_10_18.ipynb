## Frequent Words Problem

**Goal:** Find all most frequent k-mers in a given DNA string `Text`.

### Upload Input File

Please upload a text file where the first line is the DNA `Text` string and the second line is the integer `k`.

```python
from google.colab import files

def upload_and_read_input():
    uploaded = files.upload()
    for filename in uploaded.keys():
        file_content = uploaded[filename].decode('utf-8')
        print(f"Successfully loaded '{filename}'.")
        return file_content
    return None

raw_input_content = upload_and_read_input()

if raw_input_content:
    lines = raw_input_content.strip().split('\n')
    if len(lines) >= 2:
        dna_text = lines[0]
        k_value = int(lines[1])
        print(f"\nDNA Text (first 50 chars): {dna_text[:50]}...")
        print(f"K-mer length: {k_value}")
    else:
        print("Error: The uploaded file must contain at least two lines (DNA Text and k).")
        dna_text = None
        k_value = None
else:
    print("No file was uploaded or processed. Please upload a file to proceed.")
```

### `FrequentWords` Function

This function takes a `Text` (DNA string) and an integer `k` as input. It returns a list of all k-mers that appear with the highest frequency in the `Text`.

```python
def FrequentWords(Text, k):
    frequent_patterns = set()
    counts = {}
    n = len(Text)

    # Generate all k-mers and count their occurrences
    for i in range(n - k + 1):
        pattern = Text[i:i+k]
        if pattern not in counts:
            counts[pattern] = 0
        counts[pattern] += 1

    # Find the maximum count
    max_count = 0
    if counts: # Check if counts is not empty
        max_count = max(counts.values())

    # Collect all patterns that have the maximum count
    for pattern, count in counts.items():
        if count == max_count:
            frequent_patterns.add(pattern)

    return list(frequent_patterns)
```

```python
def hamming_distance(p, q):
    return sum(1 for a, b in zip(p, q) if a != b)


def reverse_complement(pattern):
    """
    Return the reverse complement of a DNA string.
    """
    complement_map = {'A': 'T', 'T': 'A', 'C': 'G', 'G': 'C'}
    complement = ''.join(complement_map[nt] for nt in pattern)
    return complement[::-1]


def neighbors(pattern, d):
    """
    Recursively generate Neighbors(Pattern, d): all k-mers whose
    Hamming distance from Pattern does not exceed d.
    """
    nucleotides = {'A', 'C', 'G', 'T'}

    if d == 0:
        return {pattern}
    if len(pattern) == 1:
        return nucleotides

    neighborhood = set()
    suffix_neighbors = neighbors(pattern[1:], d)

    for text in suffix_neighbors:
        if hamming_distance(pattern[1:], text) < d:
            for nt in nucleotides:
                neighborhood.add(nt + text)
        else:
            neighborhood.add(pattern[0] + text)

    return neighborhood


def frequent_words_with_mismatches_and_rc(text, k, d):

    freq_map = {}
    n = len(text)

    for i in range(n - k + 1):
        seq = text[i:i+k]
        neighborhood = neighbors(seq, d)
        for neighbor in neighborhood:
            freq_map[neighbor] = freq_map.get(neighbor, 0) + 1

    combined_map = {}
    for pattern in freq_map:
        rc = reverse_complement(pattern)
        combined_map[pattern] = freq_map.get(pattern, 0) + freq_map.get(rc, 0)

    max_count = max(combined_map.values())
    most_frequent = [pattern for pattern, count in combined_map.items() if count == max_count]

    return most_frequent

from google.colab import files
uploaded = files.upload()  # choose your dataset.txt file

filename = list(uploaded.keys())[0]
with open(filename) as f:
    lines = f.read().strip().split('\n')

text = lines[0].strip()
k, d = map(int, lines[1].strip().split())

result = frequent_words_with_mismatches_and_rc(text, k, d)
print(' '.join(result))
```
